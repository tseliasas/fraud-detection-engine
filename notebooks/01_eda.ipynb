## Summary: Decisions for the pipeline
1. **Metric:** PR-AUC + recall/precision. Accuracy is out, the always-"not fraud" model gets 99.8% and catches a crisp ZERO frauds.
2. **Imbalance:** 492 frauds vs 284,315 legit (~1 in 578). Needs class weights / `scale_pos_weight` or SMOTE (only on training data).
3. **Scaling:** V1–V28 are PCA'd so already ≈ 0 mean and small. Amount (max ~25,691) and Time (max ~172,792) would drown them out in logistic regression so they get scaled. XGBoost doesn't care, trees just ask "is X > something?"
4. **Time:** raw Time is useless live (seconds since some 2013 start). Valleys at ~4 AM put Time = 0 at ≈ midnight. Legit volume tanks at night but fraud doesn't, so swap Time for an **hour-of-day** feature. Caveat: only 2 days of data, some spikes might be one attacker.
5. **Duplicates:** 1,081 exact duplicate rows, 19 of them fraud (~4% of frauds). Drop them **before** splitting, otherwise copies leak into test and the score is unfair.
6. **Split:** stratified, otherwise with only ~473 frauds the test set could end up with way too few.
